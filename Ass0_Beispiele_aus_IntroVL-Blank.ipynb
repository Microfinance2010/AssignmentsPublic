#### Beispiel: Funktionen erstellen:
+ Reminder: Eine Funktion "speichert" im Prinzip ein Code-Fragment, das durch den Aufruf der Funktion ausgeführt werden kann.

Als nächstes wollen wir beim Funktionsaufruf auch noch ein "Parameter"-Wert festlegen:

+ In vielen Fällen wird durch Funktionen etwas berechnet, das wir nicht nur mit `print` "ausgegeben", sondern zur Nutzung im weiteren Programm-Verlauf AUßERHALB der Funktion einer Variablen zu weisen wollen. 
+ Dazu müssen wir das Objekt bzw. den Wert, der außerhalb der Funktion verwendet werden soll, mit `return` zurückgeben!
+ Dieses Objekt bzw. diesen Wert können wir dann einer Variable zuweisen. Der Syntax sieht wird anhand des folgenden Beispiels verdeutlicht:

```python
def fxrechner(fx, europreis):
    dollarpreis = europreis * fx
    return dollarpreis


dollarrechnung = fx_rechner(fx = 1.18, europreis = 450)

if dollarrechnung < 100:
    print("billig")
else:
    print("teuer")
```

#### Beispiel While Statements (Warnung: Schleifen lassen sich nicht elegant unterbrechen. Testen Sie also nicht extrem hohe Schleifenrunden aus!)

+ In der Programmierung entsteht oft der Bedarf, dass Code iterativ im Rahmen einer "Schleife" bzw. eines "Loops" ausgeführt würd. 
+ Die einfachste Form einer iterativen Ausführung ist ein "while"-Loop.
+ Hier wird vor jeder potentiellen Wiederholung eine Bedingung geprüft und falls die Bedigung erfüllt ist, wird der Code ausgeführt.

```python
# Wir belegen eine Variable zu Demo-Zwecken mit einem willkürlich gewählten Wert:
counter = 5
# Hier startet unser "LOOP". Die iterativ geprüfte Bedingung ist, dass die Variable counter auf einen 
# Wert größer 0 zeigt:
while counter > 0:
    # Solange die Bedingung erfüllt ist, wird folgendes gemacht:
    print ( counter )
    counter = counter - 1
print('Stop')
```

#### Beispiel Listen
+ Reminder: Neben "einfachen" Werte (z.B. Strings, Integers..) können wir auch komplexere Datenstrukturen nutzen.
+ Ein besonders wichtige Datenstruktur sind Listen, die aus verschiedene "Einträge" bestehen und im Prinzip als Container unterschiedlicher Werte betrachtet werden können..
+ Als Einträge kommen Werte in Frage, aber auch Listen selbst können einzelnen Einträgen zugewiesen werden.

```python
# Wir weisen eine neue List der Variable players zu. Die Liste enthält Strings!
players = ['Miles', 'George', 'Oscar'] 
# Wir lassen die Liste ausgeben:
print(players)
# Wir weisen einer Variable eine List mit Integers zu:
numbers = [7 , 10]
# Wir lassen die Liste ausgeben:
print(numbers)
# Die nächste Liste enhält als erstes Element einen String die anderen beiden Einträge sind selbst Listen!
mix = ["ein Eintrag", numbers, players] 
print(mix)
```

+ Wir können auf einzelne Einträge über einen "Index" zugreifen, der in eine eckige Klammer gesetzt wird.
+ Der Index automatisch indizieren, wobei der Index immer bei "0" beginnt.
+ Wir können beim Zugriff auch einen Start-Index und die End-Index angeben, wobei die beiden Größen mit Doppelpunkt getrennt werden. 

```python
mix[0]
# Wir können mehrere aufeinander folgende Einträge abfragen. Der hintere Indexwert wird dabei 
# nicht mehr mitausgegeben:
players[0:2]
# Wir ersetzen die ersten beiden Elemente der Players-Liste durch zwei neue Elemente
players[0:2] = ['Jimmy', 'Joey']
# Wir ersten ein Element durch eine Liste
numbers[1] = [99,98]
```

#### Beispiel for-Schleife
+ Wir haben bereits die while-Schleifen kennengelernt. Im Rahmen einer while-Schleife werden Bedingungen iterativ überprüft und solange die Bedingung erfüllt ist, wird ein Code iterativ ausgeführt. 
+ Im Falle von for-Schleifen "loopen" wir durch ein mehrelementiges Objekt (z.B. eine Liste) und verweisen in jedem Durchgang mit einer lokalen Variable auf das Element und führen dann ein Codestück aus, in dem auf diese lokale Variable zugegriffen werden kann:

```python
# Wir belegen zu Demo-Zwecken eine Variable mit einer beliebigen Liste:
numbers = [1 ,9 ,12]
# Am Anfang eines "For-Loops" geben wir immer eine "for-Statement" plus die Bezeichnung für den Platzhalter / 
# die lokale Variable (in dem Fall i) plus dem Wort in plus dem Element DURCH das geloopt werden soll:
for i in numbers :
    print(i+1)
```

#### Beispiel Dictionaries 

```python
dict = {'sax':'bird', 'horn': 'miles', 'guit':'george'}
print ( dict ['sax'])
len(dict)
```

+ Wir zählen die Buchstaben, die in einem String enthalten sind.
+ Dabei erzeugen wir ein Dictionary, wobei die Keys des Dictionaries jeweils die gezählten Buchstaben und der Eintrag die Anzahl sein soll:

```python
word = 'Bahnschranke' 
own_dict = {}
for letter in word:
    if letter not in own_dict: 
        own_dict[letter] = 1
    else :
        own_dict[ letter ] = own_dict[ letter ] + 1
print(own_dict)
```
