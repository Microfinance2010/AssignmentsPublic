## Vorgegebener Importcode

```python
from pathlib import Path
from openpyxl import load_workbook
arbeitsordner = Path.cwd()
dateiname = "Ass2_Reisekosten.xlsx"
dateipfad = arbeitsordner / dateiname
arbeitsmappe = load_workbook(dateipfad, data_only=True)
arbeitsblatt = arbeitsmappe["Reisekosten"]
spalten = [zelle.value for zelle in arbeitsblatt[1]]
ausgaben = []
for zeile in arbeitsblatt.iter_rows(min_row=2, values_only=True):
    ausgaben.append(dict(zip(spalten, zeile)))
print(f"{len(ausgaben)} Datensätze geladen aus: {dateipfad}")
print("Die ersten drei Datensätze:")
for datensatz in ausgaben[:3]: print(datensatz)
```
