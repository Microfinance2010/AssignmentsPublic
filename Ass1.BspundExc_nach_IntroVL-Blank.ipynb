```python
def mid(daten_liste, median=True):

    sortierte_daten = sorted(daten_liste)
    
    if median:
        if len(sortierte_daten) % 2 == 0:
            mitte_rechts = int(len(sortierte_daten) / 2)
            mitte_links = int(mitte_rechts - 1)
            mid = (sortierte_daten[mitte_links] + sortierte_daten[mitte_rechts]) / 2

        else:
            index = int(len(sortierte_daten) / 2)
            mid = sortierte_daten[index]
    else:
        mid = sum(sortierte_daten) / len(sortierte_daten)
    
    return mid

ergebnis = mid(daten_liste = [1, 9, 7, 3])
print(ergebnis)
```

#### Aufgabe: Funktion

+ Gegeben sind drei Stäbe und wir kennen deren Gesamtlänge und die Länge von zwei der drei Stäbe. 
+  Wir wissen, dass wir nur ein Dreieck aus den drei Stäben bilden können, wenn ein Stab alleine nicht größer ist als die beiden anderen Stäbe gemeinsam.
+ Schreiben Sie eine Funktion, die für eine Gesamtlänge aller Stäbe und für die Einzel-Länge für zwei der drei Stäben angibt, ob ein Dreieck aus den drei Stäben konstruiert werden kann.

```python
def dreieck(s1, s2, gesamt):
    s3 = gesamt - s1 - s2

    if s1 <= 0 or s2 <= 0 or s3 <= 0:
        return False

    if s1 >= s2 + s3 or s2 >= s1 + s3 or s3 >= s1 + s2:
        return False

    return True
```

```python
dreieck(s1=2, s2=2, gesamt=5.5)
```
